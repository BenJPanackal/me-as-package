## Pandas

I identify with the `pandas` library \[@mckinney2010\] because, like pandas, I strive to organize and clarify data. `Pandas` provides tools for manipulating structured data, and I believe in bringing structure and clarity to complex problems as well as tidyness.

``` python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Create messy raw data with missing values and inconsistencies
raw_data = {
    'Student': ['Alice', 'Bob', None, 'Diana', 'Eve'],
    'Math_Score': [85, 90, 88, None, 92],
    'Science_Score': [78, 82, 85, 88, 91],
    'GPA': [3.5, 3.8, 3.6, 3.7, 3.9]
}

df_raw = pd.DataFrame(raw_data)
print("Raw Data (with issues):")
print(df_raw)
print(f"\nMissing value counts:\n{df_raw.isnull().sum()}")

# Clean the data using pandas
df_clean = df_raw.dropna()  # Remove rows with missing values
df_clean = df_clean.reset_index(drop=True)  # Reset index
df_clean['Average_Score'] = (df_clean['Math_Score'] + df_clean['Science_Score']) / 2

print("\n\nCleaned Data:")
print(df_clean)
```

````
# Create a visualization showing the relationship between scores and GPA
plt.figure(figsize=(8, 6))
sns.scatterplot(data=df_clean, x='Average_Score', y='GPA', s=150, color='steelblue', alpha=0.7)
plt.xlabel('Average Test Score')
plt.ylabel('GPA')
plt.title('Student Performance: Test Scores vs GPA')
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
````

The visualization in @fig-student-performance shows the strong correlation between academic performance metrics, illustrating how organized data reveals patterns that raw, messy data obscures.
